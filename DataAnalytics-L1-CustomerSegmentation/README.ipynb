# Customer Segmentation Using RFM Analysis & K-Means Clustering

An unsupervised machine learning project developed as part of the **Oasis Infobyte Data Analytics Internship (Level 1, Task 2)**. This project segments a customer base by evaluating purchasing behavior using **RFM (Recency, Frequency, Monetary)** analysis combined with **K-Means Clustering**.

---

## 📌 Project Overview
Understanding customer behavior is essential for targeted marketing, personalized campaigns, and optimizing retention strategies. This project processes a transactional retail dataset to:
1. Clean and preprocess raw transaction logs.
2. Engineer **RFM (Recency, Frequency, Monetary)** metrics for individual customers.
3. Apply feature scaling and determine the optimal number of clusters using the **Elbow Method**.
4. Train a **K-Means Clustering** model to group customers into distinct behavioral segments.
5. Profile and visualize the resulting customer segments.

## 📂 Workflow & Methodology

### 1. Data Cleaning & Preprocessing
* Removed rows with missing `CustomerID`s to ensure accurate user-level aggregation.
* Filtered out canceled orders, returns, and anomalous records where `Quantity <= 0` or `UnitPrice <= 0`.
* Calculated total purchase value (`TotalPrice = Quantity * UnitPrice`) for each transaction line item.

### 2. RFM Feature Engineering
For each unique customer, three core metrics were calculated:
* **Recency (R):** Number of days elapsed since the customer's last purchase (relative to a reference snapshot date).
* **Frequency (F):** Total number of distinct transactions/invoices made by the customer.
* **Monetary (M):** Total cumulative financial spend generated by the customer.

### 3. Clustering Pipeline
* **Scaling:** Normalized the RFM feature matrix using `StandardScaler` to bring all variables to a comparable scale (mean = 0, variance = 1).
* **Elbow Method:** Evaluated Within-Cluster Sum of Squares (Inertia) across values of $K$ (from 1 to 10) to identify the optimal cluster count.
* **Model Training:** Fitted the K-Means algorithm to segment customers into actionable tiers (e.g., Champions, Loyal, At-Risk, Dormant).
