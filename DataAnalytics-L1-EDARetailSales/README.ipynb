## 🚀 Project Overview
This repository contains the complete Exploratory Data Analysis (EDA) for **Task 1** of the **Oasis Infobyte Data Analytics Internship (OIBSIP)**. The project investigates a comprehensive retail transaction dataset to uncover key performance drivers, customer purchasing behaviors, product profitability, and regional sales dynamics.

By applying rigorous data cleaning, feature engineering, and visual storytelling with Python, this project transforms raw transactional records into strategic, data-driven business insights.

---

## 🎯 Business Objectives
The analysis addresses several critical business inquiries to optimize retail operations and profitability:
1. **Temporal Sales Dynamics**: How do sales evolve across monthly, quarterly, and seasonal timelines?
2. **Customer Segmentation**: Which customer segments generate the highest revenue contribution?
3. **Product & Category Performance**: Which product categories and top-selling items drive growth versus those draining profitability through excessive discounting?
4. **Regional Disparities**: How do sales volume and net profit margins vary across different geographic regions?
5. **Discount Impact**: What is the precise relationship between discount rates and net profit margins?

---

## 📊 Dataset Overview
* **Source**: Retail Superstore Transactional Dataset
* **Dimensions**: 9,800 records $\times$ 18+ variables
* **Key Attributes**: Order Date, Ship Date, Customer Segment, Region, Category, Sub-Category, Sales, Quantity, Discount, and Profit.
